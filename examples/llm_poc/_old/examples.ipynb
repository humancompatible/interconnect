```python
import numpy as np
import pandas as pd
import json

def load_pools(path):
    data = np.load(path)
    n = int(data["num_levels"])
    pools = [data[f"level_{i}"] for i in range(n)]
    meta = json.loads(str(data["meta"])) if "meta" in data.files else {}
    return data, pools, meta
```

```python
raw, pools, meta = load_pools('pools.npz')
json.loads(str(raw['examples']))['4']
```
